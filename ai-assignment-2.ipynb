```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()

print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target labels:", digits.target_names)

plt.figure(figsize=(4, 4))
plt.imshow(digits.images[1], cmap="gray")
plt.title(f"Label: {digits.target[4]}")
plt.axis("off")
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()

fig, axes = plt.subplots(5, 2, figsize=(5, 5))

for digit, ax in enumerate(axes.flat):
    index = list(digits.target).index(digit)
    ax.imshow(digits.images[index], cmap="gray")
    ax.set_title(f"Digit: {digit}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
